# Arcadia: TripoSR on the concept pictures

Image-to-3D for the concepts made by the assets run (attached as an input). **GPU T4, Internet on.** Output: `/kaggle/working/assets/meshes/triposr/`.

```python
import os, subprocess, sys, glob, shutil
if not os.path.exists("/kaggle/working/Arcadia"):
    !git clone --depth 1 -b main https://github.com/DanielGregory/Arcadia /kaggle/working/Arcadia
%cd /kaggle/working/Arcadia
src = os.path.dirname(glob.glob("/kaggle/input/**/concepts/temple_11.png", recursive=True)[0])
os.makedirs("/kaggle/working/assets/concepts", exist_ok=True)
for f in glob.glob(f"{src}/*.png"): shutil.copy(f, "/kaggle/working/assets/concepts/")
print(sorted(os.listdir("/kaggle/working/assets/concepts")))
```

```python
subprocess.run([sys.executable, "-m", "render.make_assets", "--out", "/kaggle/working/assets", "--only", "triposr"], check=True)
shutil.rmtree("/kaggle/working/assets/_triposr_weights", ignore_errors=True)
shutil.rmtree("/kaggle/working/assets/concepts", ignore_errors=True)
```
